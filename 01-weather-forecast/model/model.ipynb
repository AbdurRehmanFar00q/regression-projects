### Model


```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import  Lasso
from sklearn.pipeline import Pipeline
from sklearn.metrics import  r2_score, mean_absolute_error
from joblib import dump, load   

df = pd.read_csv(("../data/processed/weather_features.csv"))
df['date'] = pd.to_datetime(df['date'])

x = df[[
    'temperature_2m_max', 'temperature_2m_min', 'temperature_2m_mean',
    'precipitation_sum', 'wind_speed_10m_max',
    'temp_max_lag1', 'temp_max_lag2', 'temp_max_lag7',
    'temp_max_roll_mean_7', 'temp_max_roll_std_7',
    'month', 'day_of_year', 'sin_day_of_year', 'cos_day_of_year'
]]

y = df['tomorrow_max']

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,shuffle=False)

pipeline = Pipeline([('scaler',StandardScaler()),('Lasso',Lasso(alpha=0.01))])

pipeline.fit(x_train,y_train)

y_pred = pipeline.predict(x_test)

print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))

dump(pipeline, '../model/lasso_model.joblib')



```
